```python
!python -m spacy download en_core_web_sm
import spacy 
nlp = spacy.load('en_core_web_sm')
```

```python
with open("peterrabbit.txt", "r") as f:
    text = f.read()
doc = nlp(text)
```

```python
#For every token in the third sentence, print the token text, the POS tag, the fine-grained TAG tag, and the description of the fine-grained tag.
sentence = list(doc.sents)
third_sentence = sentence[2]
for token in third_sentence:
    print(token.text,token.pos_,token.tag_,spacy.explain(token.tag_))
```

```python
#Provide a frequency list of POS tags from the entire document
from collections import Counter
pos_freq = Counter()
for token in doc:
    pos_freq[token.pos_] += 1
print(pos_freq)
```

```python
#CHALLENGE: What percentage of tokens are nouns?
noun_count = 0
total_tokens = 0
for token in doc:
    if token.is_alpha:
        total_tokens += 1
        if token.pos_ == "NOUN":
            noun_count += 1
percentage = (noun_count / total_tokens) * 100
print(f"{percentage:.2f}%")
```

```python
#Display the Dependency Parse for the third sentence.
from spacy import displacy
third_sentence = list(doc.sents)[2]
displacy.render(third_sentence, style="dep", jupyter=True)
```

```python
#Show the first two named entities from Beatrix Potter's The Tale of Peter Rabbit
for ent in doc.ents[:2]:
    print(ent.text, ent.label_)
```

```python
#How many sentences are contained in The Tale of Peter Rabbit?
num_sentences = len(list(doc.sents))
print(num_sentences)
```

```python
#CHALLENGE: How many sentences contain named entities?
count = 0
for sent in doc.sents:
    if len(sent.ents) > 0:
        count += 1
print(count)
```
